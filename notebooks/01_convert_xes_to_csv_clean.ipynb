```python
log = pm4py.read_xes("BPI_Challenge_2012.xes.gz")
df = pm4py.convert_to_dataframe(log)
```

```python
df.head()
```

```python
df.columns
```

```python
df_clean = df.rename(columns={
    "org:resource": "resource",
    "lifecycle:transition": "lifecycle_transition",
    "concept:name": "activity",
    "time:timestamp": "timestamp",
    "case:REG_DATE": "case_registration_date",
    "case:concept:name": "case_id",
    "case:AMOUNT_REQ": "amount_requested"
})
```

```python
df_clean.info()
```

```python
df_clean.to_csv("bpi_2012_events.csv", index=False)
```

```python
df_sample = df_clean.sample(1000, random_state=42)
```

```python
df_sample.to_csv("sample_bpi_2012_events.csv", index=False)
```
